```python
import numpy as np
import matplotlib.pyplot as plt

metrics = ["IL1 Miss", "UL2 Miss", "Speed"]

# 실험 선택 (예: B-1way, B-2way, C-128B, D-LRU, E-2)
selected_exp = ["B-1way", "B-2way", "C-128B", "D-LRU", "E-Unified"]

data = [
    [8985499, 2665407, 11111111],
    [503,     2653138, 10526315],
    [503,     2653138, 7246376],
    [503,     2653138, 10989010],
    [503,     2449462, 11111111]
]

# 정규화
data_norm = np.array(data) / np.max(data, axis=0)

angles = np.linspace(0, 2*np.pi, len(metrics), endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(8,8))

for i, label in enumerate(selected_exp):
    values = data_norm[i].tolist()
    values += values[:1]
    plt.plot(angles, values, label=label)
    plt.fill(angles, values, alpha=0.2)

plt.xticks(angles[:-1], metrics)
plt.title("Overall Performance Radar Chart")
plt.legend(loc='upper right')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# -----------------------------
# 대표 실험 이름
# -----------------------------
labels = ["A-64KB", "B-1way", "B-2way", "C-32B", "C-64B", "D-LRU", "E-Split", "E-Unified"]

# -----------------------------
# 각 실험의 값 (IL1 miss, UL2 miss, sim speed)
# -----------------------------
data = [
    [503, 2653469, 10638297],     # A-64KB
    [8985499, 2665407, 11111111], # B-1way
    [503, 2653138, 10526315],     # B-2way
    [1574, 2653195, 8403361],     # C-32B
    [882, 2653167, 10638297],     # C-64B
    [503, 2653138, 10989010],     # D-LRU
    [5373562, 5373562, 10989010], # E-Split (huge UL2 miss)
    [503, 2449462, 11111111]      # E-Unified
]

# -----------------------------
# 정규화 (0~1 범위)
# -----------------------------
data = np.array(data)
norm = data / data.max(axis=0)

# -----------------------------
# Radar Chart 설정
# -----------------------------
metrics = ["IL1 Miss", "UL2 Miss", "Speed"]
angles = np.linspace(0, 2*np.pi, len(metrics), endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(10,10))

for i in range(len(labels)):
    values = norm[i].tolist()
    values += values[:1]
    plt.plot(angles, values, linewidth=2, label=labels[i])
    plt.fill(angles, values, alpha=0.15)

plt.xticks(angles[:-1], metrics, fontsize=13)
plt.yticks(fontsize=11)
plt.title("Final Unified Performance Comparison (A–E Experiments)", fontsize=16)
plt.legend(loc="upper right", bbox_to_anchor=(1.35, 1.15))
plt.grid(True)
plt.show()
```
